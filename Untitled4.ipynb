```python
import pandas as pd
import matplotlib.pyplot as plt

work = df.copy()

date_col = "release_date" if "release_date" in work.columns else "release_year"
rating_col = "vote_average" if "vote_average" in work.columns else "rating"
votes_col = "vote_count" if "vote_count" in work.columns else None

work["year"] = pd.to_datetime(work[date_col], errors="coerce").dt.year
if work["year"].isna().all():
    work["year"] = pd.to_numeric(work[date_col], errors="coerce")

work = work.dropna(subset=["year", rating_col])
work["year"] = work["year"].astype(int)
work["decade"] = (work["year"] // 10) * 10

if votes_col:
    work = work[work[votes_col] >= 100]

decade_stats = (
    work.groupby("decade")
    .agg(
        avg_rating=(rating_col, "mean"),
        movie_count=(rating_col, "size"),
    )
    .reset_index()
)

decade_stats
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(decade_stats["decade"], decade_stats["avg_rating"], marker="o")
ax.set_title("Average TMDB movie rating by decade")
ax.set_xlabel("Decade")
ax.set_ylabel("Average rating")
ax.grid(True, alpha=0.3)
plt.show()
```

```python
rating_col = "vote_average" if "vote_average" in df.columns else "rating"
votes_col = "vote_count" if "vote_count" in df.columns else None
date_col = "release_date" if "release_date" in df.columns else "release_year"

work = df.copy()
work["year"] = pd.to_datetime(work[date_col], errors="coerce").dt.year
if work["year"].isna().all():
    work["year"] = pd.to_numeric(work[date_col], errors="coerce")
work = work.dropna(subset=["year", rating_col])
work["year"] = work["year"].astype(int)

if votes_col:
    work = work[work[votes_col] >= 100]

# "Top rated" = 7.5 or higher
top_rated = work[work[rating_col] >= 7.5]

by_year = (
    top_rated.groupby("year")
    .size()
    .reset_index(name="top_rated_count")
    .sort_values("top_rated_count", ascending=False)
)

print("Year with the most top-rated movies:")
print(by_year.head(1).to_string(index=False))

print("\nTop 10 years:")
print(by_year.head(10).to_string(index=False))

fig, ax = plt.subplots(figsize=(12, 5))
ax.bar(by_year["year"], by_year["top_rated_count"])
ax.set_title("Number of movies rated 7.5+ by year")
ax.set_xlabel("Year")
ax.set_ylabel("Count of top-rated movies")
plt.tight_layout()
plt.show()
```
